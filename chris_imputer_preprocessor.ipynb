```python
import pandas as pd
```

```python
# MIQ — Feature Imputer for Preprocessor

# MUST HAVE — user always inputs, no imputation needed
MUST_HAVE = [
    'age',
    'running_experience_months',
    'weekly_mileage_km',
    'injury_count',
    'injury_severity',
    'course_difficulty',
]

# NICE TO HAVE — user inputs OR preprocessor fills

# a. Median imputation — continuous health metrics
NICE_TO_HAVE_MEDIAN = [
    'vo2_max',
    'resting_heart_rate_bpm',
    'recovery_score',
    'nutrition_score',
]

# b. Zero imputation — count/behavioural features
NICE_TO_HAVE_ZERO = [
    'previous_marathon_count',      # 0 = first timer
    'run_club_attendance_rate',     # 0 = no club
]

# c. Neutral benchmark — weather (user-selectable, defaults to neutral)
NICE_TO_HAVE_WEATHER = {
    'marathon_weather_Cold':  0,    # neutral = not cold
    'marathon_weather_Hot':   0,    # neutral = not hot
    'marathon_weather_Rainy': 0,    # neutral = not rainy
    'marathon_weather_Windy': 0,    # neutral = not windy
}

# Full feature set = MUST_HAVE + all NICE_TO_HAVE
ALL_FEATURES = (
    MUST_HAVE +
    NICE_TO_HAVE_MEDIAN +
    NICE_TO_HAVE_ZERO +
    list(NICE_TO_HAVE_WEATHER.keys())
)

print(f"Total features: {len(ALL_FEATURES)}")
print(ALL_FEATURES)
```

```python
# Imputation function

def impute_missing(df_input: pd.DataFrame, 
                   feature_medians: pd.Series) -> pd.DataFrame:
    df = df_input.copy()

    # a. Median imputationwww
    for col in NICE_TO_HAVE_MEDIAN:
        if col in df.columns:
            df[col] = df[col].fillna(feature_medians[col])

    # b. Zero imputation
    for col in NICE_TO_HAVE_ZERO:
        if col in df.columns:
            df[col] = df[col].fillna(0)

    # c. Weather — neutral benchmark
    for col, default in NICE_TO_HAVE_WEATHER.items():
        if col in df.columns:
            df[col] = df[col].fillna(default)

    return df
```
